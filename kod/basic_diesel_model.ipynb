# Pouze stanice

```python
df_stations = df_rizeny_obd.select(
    pl.col('StaniceCislo'),
    pl.col('Vysledek_Vyhovuje')
)
df_stations
```

```python
# Celkový průměr a celková variabilita (SST)
total_mean = df_stations.select(pl.col("Vysledek_Vyhovuje").mean()).item()
df_with_sq_diff = df_stations.with_columns(
    ((pl.col("Vysledek_Vyhovuje").cast(pl.Float64) - total_mean)**2).alias("sq_diff")
)
sst = df_with_sq_diff.select(pl.sum("sq_diff")).item()

# Variabilita uvnitř skupin (stanic) (SSE)
# Spočítáme průměr pro každou stanici a reziduální součet čtverců
sse = (
    df_stations.group_by("StaniceCislo")
    .agg([
        pl.col("Vysledek_Vyhovuje").mean().alias("avg_stanice"),
        pl.col("Vysledek_Vyhovuje").count().alias("n_stanice"),
        pl.col("Vysledek_Vyhovuje").var().alias("var_stanice")
    ])
    .select(
        # SSE = suma((n-1) * variance)
        ((pl.col("n_stanice") - 1) * pl.col("var_stanice")).sum()
    )
    .item()
)

# Výpočet R-squared
r_squared = 1 - (sse / sst)

print(f"Stanice vysvětlují {r_squared * 100:.4f} % variability výsledku.")
```

```python
# Výpočet úspěšnosti pro každou stanici
station_success = (
    df_stations.with_columns(
        pl.col("Vysledek_Vyhovuje").cast(pl.Float64)
    )
    .group_by("StaniceCislo")
    .agg([
        pl.col("Vysledek_Vyhovuje").mean().alias("Procento_Uspechu"),
        pl.col("Vysledek_Vyhovuje").count().alias("Pocet_Kontrol")
    ])
    .sort("Procento_Uspechu", descending=True)
)

display(station_success)
```

```python
# 1. Filtrace a příprava
df_plot = (
    station_success
    .filter(pl.col("Pocet_Kontrol") >= 100)
    .to_pandas()
)

# 2. Tvorba grafu (bez marginal_y pro stabilitu se spojitou barvou)
fig = px.scatter(
    df_plot,
    x="Pocet_Kontrol",
    y="Procento_Uspechu",
    color="Procento_Uspechu",
    size="Pocet_Kontrol",
    hover_name="StaniceCislo",
    log_x=True,
    range_y=[0.7, 1.01],
    title="Úspěšnost STK (stanice s více než 100 kontrolami)",
    labels={
        "Pocet_Kontrol": "Počet kontrol (log osa)",
        "Procento_Uspechu": "Míra úspěšnosti"
    },
    template="plotly_dark",
    color_continuous_scale="RdYlGn" 
)

# Nastavení velikosti bodů, aby nebyly příliš malé/velké
fig.update_traces(marker=dict(sizeref=2.*max(df_plot['Pocet_Kontrol'])/(40.**2), sizemode='area'))

fig.show()
```
